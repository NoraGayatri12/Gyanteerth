```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------- OOPS ----------
class Vehicle:
    def __init__(self, vehicle_id, vehicle_type, price_per_day):
        self.vehicle_id = vehicle_id
        self.vehicle_type = vehicle_type
        self.price_per_day = price_per_day

    def calculate_price(self, days):
        return self.price_per_day * days


# ---------- Vehicles ----------
car = Vehicle(1, "Car", 1500)
bike = Vehicle(2, "Bike", 500)

vehicles = {1: car, 2: bike}
rental_records = []

# ---------- Rental Process ----------
for i in range(3):  # Sample rentals
    print("\n1. Car\n2. Bike")
    choice = int(input("Choose Vehicle: "))
    days = int(input("Enter rental days: "))

    vehicle = vehicles[choice]
    cost = vehicle.calculate_price(days)

    rental_records.append({
        "Vehicle": vehicle.vehicle_type,
        "Days": days,
        "Cost": cost
    })

    print(f"Total Cost: ₹{cost}")

# ---------- Pandas DataFrame ----------
df = pd.DataFrame(rental_records)
print("\nRental Summary:\n", df)

# ---------- NumPy Analysis ----------
total_revenue = np.sum(df["Cost"])
average_rent = np.mean(df["Cost"])

print("\nTotal Revenue:", total_revenue)
print("Average Rental Cost:", average_rent)

# ---------- Visualization ----------
df.groupby("Vehicle")["Cost"].sum().plot(kind="bar")
plt.title("Revenue by Vehicle Type")
plt.xlabel("Vehicle")
plt.ylabel("Revenue")
plt.show()
```
