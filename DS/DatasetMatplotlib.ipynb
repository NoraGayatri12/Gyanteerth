```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_excel('Sales Data.xlsx')
df.head()
```

```python
product_sales=df.groupby("Product")["TotalPrice"].min()
print(product_sales)
```

```python
plt.figure(figsize=(10,5))
plt.bar(product_sales.index,product_sales.values)
plt.title('Daily Total Sales Trend')
plt.xlabel('Product')
plt.ylabel('Total price')


```

```python
#agg multiple aggregation
product_sales=df.groupby(["Product",'PaymentMethod'])['TotalPrice].agg(['sum','min','max'])
```

```python
product_sales=df.groupby("Product")["TotalPrice"].min()
print(product_sales)
plt.plot(product_sales)
plt.xlabel("Product")
plt.ylabel("Total Price")
```
