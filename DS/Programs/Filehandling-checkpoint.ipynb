## Write operations

```python
S1=open("DataS.txt","w")
S1.write("I am Gayatri")
S1.close()
```

```python
S1=open("DataS.txt","w")
S1.writelines(["I am Gayatri\n","I am From Solapur\n","Thank you!"])
S1.close()
```

## Append (No overwrite)

```python
S1=open("DataS.txt","a")
S1.write("\nThank You!!!")
S1.close()
```

```python
S1=open("DataS.txt","r+")
print(S1.read())
S1.write("\nStay Positive")
S1.close()
```

```python
S1=open("DataS.txt","w+")
S1.write("\nGood Bye")
print(S1.read())
S1.close()
```
