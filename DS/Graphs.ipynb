## BFS

```python
from collections import deque
graph={
    'A':['B','C'],
    'B':['D'],
    'C':['E'],
    'D':[],
    'E':[]
}
def bfs(graph,start):
    visited=set()
    queue=deque([start])
    while queue:
        node=queue.popleft()
        if node not in visited:
            print(node,end=" ")
            visited.add(node)
            queue.extend(graph[node])
bfs(graph,'A')
```
