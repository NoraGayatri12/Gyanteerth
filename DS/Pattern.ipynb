```python
n=4
for i in range(n):
    for j in range(n):
        print("*",end=" ")
    print()
```

```python
n=4
for i in range(1,n+1):
    for j in range(i):
        print("*",end=" ")
    print()
```

```python
n=4
for i in range(n,0,-1):
    for j in range(i):
        print("*",end=" ")
    print()
```

```python
#pyramid pattern
n=4
for i in range(1,n+1):
    print(" "*(n-i),end="")
    print("* "*i)
```

```python
n=4
for i in range(n,0,-1):
    print(" "*(n-i),end="")
    print("* "*i)
```
