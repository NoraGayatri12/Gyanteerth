```python
import seaborn as sns
```

```python
import pandas as pd
d=pd.read_excel("dataset2.xlsx")
print(d)
```

```python
sns.relplot(x='Hours',y='marks',hue='age',data=d)
```

```python
sns.relplot(x='Hours',y='marks',hue='age',data=d,kind="line",style="internet")
```

```python
sns.displot(d,x='Hours',y='marks',hue="internet")
```

```python
sns.relplot(x='Hours',y='marks',hue='age',data=d,col="internet")
```

```python
sns.displot(d,x='age',binwidth=5,hue="internet")
```

```python
sns.regplot(x='marks',y="Hours",data=d)
```

```python
sns.displot(d,x='marks',col='internet')
```

```python
sns.displot(d,x='age',y='marks',hue='internet')
```

```python
sns.catplot(d,x='age',y='marks',hue='internet')
```

```python
sns.catplot(x='age',y='marks',kind="box",data=d,hue='internet')
```
