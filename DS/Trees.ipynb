## Binary Search Tree

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
        
def insert(root,key):
    if root is None:
        return Node(key)
    if key<root.data:
        root.left=insert(root.left,key)
    else:
        root.right=insert(root.right,key)
    return root
def add_inorder(self):
    if self:
        add_inorder(self.left)
        print(self.data,end=" ")
        add_inorder(self.right)
#Main
root=None
for i in [50,20,30,10,60,40,30]:
    root=insert(root,i)
print("sorted output:")
add_inorder(root)
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data,end=" ")
        inorder(root.right)
def preorder(root):
    if root:
        print(root.data,end=" ")
        preorder(root.left)
        preorder(root.right)
def postorder(root):
    if root:
        postorder(root.left)
        print(root.data,end=" ")
        postorder(root.right)
root=None
for i in [10,30,20,60,50,40]:
    root=insert(root,i)
print("\ninorder")
inorder(root)
print("\npreorder")
preorder(root)
print("\npostorder")
preorder(root)
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data,end=" ")
        inorder(root.right)
def preorder(root):
    if root:
        print(root.data,end=" ")
        preorder(root.left)
        preorder(root.right)
def postorder(root):
    if root:
        postorder(root.left)
        print(root.data,end=" ")
        postorder(root.right)
root=Node(10)
root.left=Node(5)
root.right=Node(20)
root.left.left=Node(3)
root.right.right=Node(7)
print("\ninorder:")
inorder(root)
print("\npreorder:")
preorder(root)
print("\npostorder:")
preorder(root)
```
