```python
import pandas as pd
import matplotlib.pyplot as plt

class Vehicle:
    def __init__(self, vehicle_id, vehicle_type, price):
        self.vehicle_id = vehicle_id
        self.vehicle_type = vehicle_type
        self.price = price

    def calculate_price(self, days):
        cost=self.price * days
        return cost



bike = Vehicle(1, "Bike", 1000)
car = Vehicle(2, "Car", 5000)

vehicles = {1: bike, 2: car}
rental_records = []


while True:
    print("\n1. Rent Vehicle")
    print("2. Show Revenue")
    print("3. Exit")

    a = int(input("Choose Option: "))

    
    if a == 1:
        print("1. Bike\n2. Car")
        choose = int(input("You choose: "))
        days = int(input("Enter Rental Days: "))

        vehicle = vehicles[choose]
        cost = vehicle.calculate_price(days)

        rental_records.append({
            "Vehicle": vehicle.vehicle_type,
            "Days": days,
            "Cost": cost
        })

        print(f"Total Cost: ₹{cost}")

    
    elif a == 2:
        if not rental_records:
            print("No data available!")
        else:
            df = pd.DataFrame(rental_records)
            print(df)

            df.groupby("Vehicle")["Cost"].sum().plot(kind="bar")
            plt.title("Revenue by Vehicle Type")
            plt.xlabel("Vehicle")
            plt.ylabel("Revenue")
            plt.show()

    
    elif a == 3:
        print("Thank you!")
        break

    else:
        print("Invalid option! Try again.")
```
