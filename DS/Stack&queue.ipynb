## Queue

```python
from collections import deque
Queue=deque()
Queue.append("P1")
Queue.append("P2")
Queue.append("P3")
print("After action",Queue)
last_action=Queue.popleft() #pop from first
print("After action",last_action)
```

```python
class Browser:
    def __init__(self):
        self.history=[]
    def open_page(self,page):
        self.history.append(page)
        print(page,"opened")
    def back(self):
        if self.history:
            print("Back from",self.history.pop())
        else:
            print("No pages")
b=Browser()
b.open_page("Google")
b.open_page("Youtube")
b.open_page("Gmail")
b.back()
```

```python
class CallCentre:
    def __init__(self):
        self.calls=[]
    def receive_call(self,call_id):
        self.calls.append(call_id)
        print("call received",call_id)
    def attend_call(self):
        if self.calls:
            print("Attending call:",self.calls.pop(0))
        else:
            print("No Calls")
c=CallCentre()
c.receive_call(101)
c.receive_call(102)
c.receive_call(103)
c.attend_call()
```

```python
class CallCentre:
    def __init__(self):
        self.calls=deque()
    def receive_call(self,call_id):
        self.calls.append(call_id)
        print("call received",call_id)
    def attend_call(self):
        if self.calls:
            print("Attending call:",self.calls.popleft())
        else:
            print("No Calls")
c=CallCentre()
c.receive_call(101)
c.receive_call(102)
c.receive_call(103)
c.attend_call()
```

## Linked List

```python
class Person:
    def __init__(self,name):
        self.name=name
        self.next=None
a=Person("Gayatri")
b=Person("Devidas")
c=Person("Nora")
a.next=b
b.next=c
#connecting
temp=a
while temp:
    print(temp.name,end="->")
    temp=temp.next
print("None")
```

```python
#reverse list using stack
list=[10,20,30,40]
stack=[]
for i in list:
    stack.append(i)
print("original list",stack)
reverse_list=[]
while stack:
    reverse_list.append(stack.pop())
print("reverse list",reverse_list)
    
```

#### Ticket Booking System using Queue

```python
from collections import deque
Queue=deque()
while True:
    print("\n1.Booking Ticket")
    print("2.Serve Ticket")
    print("3.Waiting list")
    print("4.Exit")
    ch=int(input("Enter your Choice"))
    if ch==1:
        Name=input("Enter Customer Name:")
        Queue.append(Name)
        print("Customer coming in queue",Name)
    elif ch==2:
        if Queue:
            Customer= Queue.popleft()
            print("Ticket Booking for",Customer)
        else:
            print("No in queue")
    elif ch==3:
        if Queue:
            print("Waiting in a Queue",list(Queue))
        else:
            print("Queue is Empty")
        
    elif ch==4:
        print("Thank You!")
        break;
    else:

        
        print("Enter Above Choice")
        
        
```
