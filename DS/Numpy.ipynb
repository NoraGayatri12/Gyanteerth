```python
import numpy as np
```

```python
a=np.array([10,20,30,40,50])
print(a)
print(a.ndim)
```

```python
a=np.array([[10,20,30,40,50]])
print(a)
print(a.ndim)
```

```python
# Range in array
np.arange(10)
```

```python
c=np.arange(10)
d=np.linspace(1,10,10)
print(c)
print(d)
```

```python
c=np.arange(10)
d=np.linspace(1,5,6)
print(c)
print(d)
```

```python
a=np.array([7,9,9,4])
b=np.array([(3,6,9),(11,22,33)])
print(a.ndim) # dimention of array
print(a.itemsize)# byte size of each element
print(a.size*a.itemsize)# memory occupied-space occupide
print(a.size)#no. of element in array
print(b.shape)# shape of array(rows and columns)
```

```python
z=np.zeros(4)
print(z)
```

```python
z=np.ones(5)
print(z)
```

```python
#Identity Matrix
z=np.eye(4)
print(z)
```

```python
z=np.full(5,2)#5 element print with 2
print(z)
```

```python
z=np.arange(1,10,2)# arange(start,end,step)
print(z)
```

```python
#run random data or value
z=np.random.rand(10)
print(z)
```

```python
#run random data or value within given range
z=np.random.randint(1,100,20)#(start,end,elements)
print(z)
```
